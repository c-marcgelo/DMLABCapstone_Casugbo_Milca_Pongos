```python

supply = pd.read_csv(
    "Physical Water Supply.csv",
    sep=";",
    skiprows=2,
    na_values=[".", ".."]
)

use = pd.read_csv(
    "Physical Water Use.csv",
    sep=";",
    skiprows=2,
    na_values=[".", ".."]
)

withdrawal = pd.read_csv(
    "industrial-water-withdrawal.csv"
)

water_accounts = pd.ExcelFile(
    "FINAL - 26ENREAS01-114 - 2025 Water Accounts Web Release Materials - Web Release Materials - 4 of 4.xlsx"
)

print("All datasets loaded successfully!")
```

```python

datasets = {
    "Physical Water Supply": supply,
    "Physical Water Use": use,
    "Industrial Water Withdrawal": withdrawal
}

for name, df in datasets.items():
    print("\nDataset:", name)
    print("Number of observations:", df.shape[0])
    print("Number of attributes:", df.shape[1])
    print("Columns:", list(df.columns))

print("\nWater Accounts Excel Workbook")
print("Number of worksheets:", len(water_accounts.sheet_names))
print("Worksheet names:")
for sheet in water_accounts.sheet_names:
    print("-", sheet)
```

```python
for name, df in datasets.items():
    print("\n", "=" * 60)
    print("DATASET:", name)
    display(df.head())
```

**Data Types and Attributes**

```python
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    print("\nDATA TYPES:")
    print(df.dtypes)

    print("\nDATA INFORMATION:")
    df.info()
```

```python
for sheet in ["3", "4"]:
    print("\n" + "=" * 60)
    print("WORKSHEET:", sheet)
    print("=" * 60)

    df = pd.read_excel(water_accounts, sheet_name=sheet)

    print("Dimensions:", df.shape)
    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    display(df.head())
```

```python
print("PHYSICAL WATER SUPPLY")
print("=" * 50)

print("\nDATA TYPES:")
print(supply.dtypes)

print("\nDATA INFORMATION:")
supply.info()
```

**Missing Values**

```python
for name, df in datasets.items():

    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    missing = df.isnull().sum()

    missing_summary = pd.DataFrame({
        "Missing Values": missing,
        "Percentage": (missing / len(df)) * 100
    })

    display(missing_summary)
```

```python
for sheet in ["3", "4"]:

    print("\n" + "=" * 60)
    print("WORKSHEET:", sheet)
    print("=" * 60)

    df = pd.read_excel(
        water_accounts,
        sheet_name=sheet
    )

    print("Missing values per column:")
    display(df.isnull().sum())

    print("\nTotal missing values:", df.isnull().sum().sum())
```

**Duplicate Records**

```python
for name, df in datasets.items():

    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    duplicate_count = df.duplicated().sum()

    print("Total duplicate records:", duplicate_count)

    print(
        "Percentage of duplicates:",
        round((duplicate_count / len(df)) * 100, 2),
        "%"
    )
```

```python
philippines = withdrawal[
    withdrawal["Code"] == "PHL"
]

print("Philippine records:", len(philippines))
print(
    "Duplicate Philippine records:",
    philippines.duplicated().sum()
)

print(
    "Duplicate years:",
    philippines["Year"].duplicated().sum()
)
```

**Invalid and Unusual Values**

```python
for name, df in datasets.items():

    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    print("\nNUMERICAL SUMMARY:")
    display(df.describe())

    print("\nNEGATIVE VALUES:")
    numeric_df = df.select_dtypes(include=["number"])

    print((numeric_df < 0).sum())
```

```python
for name, df in datasets.items():

    print("\nDATASET:", name)

    print("Minimum Year:", df["Year"].min())
    print("Maximum Year:", df["Year"].max())

    print("Unique Years:", df["Year"].nunique())
```

```python
philippines = withdrawal[
    withdrawal["Code"] == "PHL"
].copy()

print("PHILIPPINES INDUSTRIAL WATER WITHDRAWAL")
print("=" * 50)

display(
    philippines[
        ["Year", "Industrial water withdrawal"]
    ].describe()
)

print("\nNegative values:")
print(
    (philippines["Industrial water withdrawal"] < 0).sum()
)

print("\nZero values:")
print(
    (philippines["Industrial water withdrawal"] == 0).sum()
)
```

```python
for name, df in datasets.items():

    numeric_df = df.select_dtypes(include=["number"])

    print("\nDATASET:", name)
    print("=" * 50)

    print("Negative values per column:")
    print((numeric_df < 0).sum())

    print("\nZero values per column:")
    print((numeric_df == 0).sum())

    print("\nMinimum values:")
    print(numeric_df.min())

    print("\nMaximum values:")
    print(numeric_df.max())
```

**Categorical and Numerical Variables**

```python
for name, df in datasets.items():

    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    categorical = df.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    numerical = df.select_dtypes(
        include=["number"]
    ).columns.tolist()

    print("\nCategorical Variables:")
    for col in categorical:
        print("-", col)

    print("\nNumerical Variables:")
    for col in numerical:
        print("-", col)
```

```python
philippines = withdrawal[
    withdrawal["Code"] == "PHL"
].copy()

print("PHILIPPINES DATASET")
print("\nCategorical Variables:")
print(
    philippines.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()
)

print("\nNumerical Variables:")
print(
    philippines.select_dtypes(
        include=["number"]
    ).columns.tolist()
)
```
